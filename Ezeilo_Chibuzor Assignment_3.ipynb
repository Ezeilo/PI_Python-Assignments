```python
purchase_amount = float(input("Enter purchase total: "))
if purchase_amount >= 10000:
    discount = 0.15
elif purchase_amount >= 5000:
    discount = 0.10
elif purchase_amount >= 2000:
    discount = 0.05
elif purchase_amount < 2000:
    discount = 0
else:
    print("Error")
print(f"Your discount is {discount * 100}%")
```
